# **Q1. `nn.Module`**

**Understand the Concept**

- `nn.Module` is the base class used to build PyTorch neural networks.
- We create our model as a class that inherits from it.
- It keeps track of layers and trainable parameters.

```python
import torch.nn as nn

class SimpleNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer = nn.Linear(2, 3)

model = SimpleNetwork()

print(model)
```

# **Explanation**

- The model inherits from `nn.Module`.
- `super().__init__()` initializes the PyTorch module.
- The layer becomes part of the model.

# **Q2. Constructor**

**Understand the Concept**

- The constructor is the `__init__()` method.
- We define the layers and model components inside it.
- `super().__init__()` initializes the parent `nn.Module`.

```python
import torch.nn as nn

class Network(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(4, 8)
        self.fc2 = nn.Linear(8, 1)

model = Network()

print(model)
```

# **Explanation**

- `fc1` and `fc2` are created inside the constructor.
- These layers become part of the model architecture.

# **Q4. `nn.Linear`**

**Understand the Concept**

- `nn.Linear` creates a fully connected layer.
- It performs a linear transformation.

**Formula**

`y = xWᵀ + b`

**Arguments**

`nn.Linear(input_features, output_features)`

```python
import torch
import torch.nn as nn

layer = nn.Linear(3, 2)

x = torch.tensor([[1.0, 2.0, 3.0]])
output = layer(x)

print("Output:", output)
print("Weight shape:", layer.weight.shape)
print("Bias shape:", layer.bias.shape)
```

# **Explanation**

- The layer accepts 3 input features.
- It produces 2 output values.
- The weight matrix and bias vector are trainable parameters.

# **Q5. Forward Method**

**Understand the Concept**

- The `forward()` method defines how input moves through the network.
- When we call `model(x)`, PyTorch automatically calls `forward()`.

```python
import torch
import torch.nn as nn

class Network(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(2, 1)

    def forward(self, x):
        return self.fc(x)

model = Network()

x = torch.tensor([[2.0, 3.0]])
print(model(x))
```

# **Explanation**

- `forward()` defines the calculation.
- Calling `model(x)` runs the forward method automatically.

# **Q9. Build a Complete Neural Network**

**Understand the Concept**

Network structure:

`Input → Hidden Layer → Activation → Hidden Layer → Output`

```python
import torch
import torch.nn as nn

class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(4, 8)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(8, 6)
        self.output = nn.Linear(6, 2)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        x = self.relu(x)
        x = self.output(x)
        return x

model = NeuralNetwork()

print(model)
```

# **Explanation**

- The model has 4 input features.
- The first hidden layer has 8 neurons.
- The second hidden layer has 6 neurons.
- The output layer has 2 neurons.
- ReLU is used between the layers.

# **Q11. Model Parameters**

**Understand the Concept**

- Parameters are values learned during training.
- They mainly include weights and biases.
- The number of parameters depends on the layer sizes.

```python
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(4, 8),
    nn.ReLU(),
    nn.Linear(8, 2)
)

total = sum(p.numel() for p in model.parameters())

print("Total parameters:", total)
```

# **Explanation**

- `numel()` counts the values inside each parameter tensor.
- Adding them gives the total number of model parameters.

# **Q12. Trainable Parameters**

**Understand the Concept**

- Trainable parameters are parameters whose values can be updated during training.
- In PyTorch, `requires_grad=True` normally means the parameter can receive gradients.

```python
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(4, 8),
    nn.ReLU(),
    nn.Linear(8, 2)
)

trainable = sum(
    p.numel() for p in model.parameters() if p.requires_grad
)

print("Trainable parameters:", trainable)
```

# **Explanation**

- The code counts only parameters with `requires_grad=True`.
- These are the parameters the optimizer can update.

# **Q13. Weight Matrices and Bias Vectors**

**Understand the Concept**

- Each `nn.Linear` layer has a weight matrix and bias vector.
- Their shapes depend on the input and output features.

```python
import torch.nn as nn

layer = nn.Linear(4, 8)

print("Weight shape:", layer.weight.shape)
print("Bias shape:", layer.bias.shape)

print("\nWeights:")
print(layer.weight)

print("\nBias:")
print(layer.bias)
```

# **Explanation**

- For `nn.Linear(4, 8)`, the weight has 8 rows and 4 columns.
- The bias has 8 values.
- These values are learned during training.

# **Q14. Inspect Model Architecture**

**Understand the Concept**

- Printing the model shows its architecture.
- We can inspect layers and parameter counts before training.

```python
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(4, 8),
    nn.ReLU(),
    nn.Linear(8, 6),
    nn.ReLU(),
    nn.Linear(6, 2)
)

print(model)

print("\nParameter count:")
print(sum(p.numel() for p in model.parameters()))
```

# **Explanation**

- Printing `model` shows the layer structure.
- The parameter count shows how many values the model can learn.
