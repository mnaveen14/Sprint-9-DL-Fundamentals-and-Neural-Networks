## **4. Tensor vs NumPy Array**

**Understand the Concept**

## **5. Creating Tensors**

**Understand the Concept**

- PyTorch provides different ways to create tensors.
- We can create tensors from lists, NumPy arrays, or fixed values.
- The way we create a tensor depends on what data we need.

**How It Works**

- `torch.tensor()` creates a tensor from existing values.
- `torch.zeros()` creates a tensor filled with zeros.
- `torch.ones()` creates a tensor filled with ones.
- `torch.rand()` creates a tensor with random values.



```python
import torch

# From list
tensor1 = torch.tensor([10, 20, 30])

# Zeros
tensor2 = torch.zeros(2, 3)

# Ones
tensor3 = torch.ones(2, 3)

# Random values
tensor4 = torch.rand(2, 3)

print("From list:")
print(tensor1)

print("\nZeros:")
print(tensor2)

print("\nOnes:")
print(tensor3)

print("\nRandom:")
print(tensor4)
```

## **7. Tensor Dimensions**

**Understand the Concept**

- Tensor dimension tells us how many axes a tensor has.
- A 0D tensor contains a single value.
- A 1D tensor contains a list of values.
- A 2D tensor contains rows and columns.



```python
# Tensor Dimensions
# 0D tensor
tensor_0d = torch.tensor(10)

# 1D tensor
tensor_1d = torch.tensor([10, 20, 30])

# 2D tensor
tensor_2d = torch.tensor([[10, 20],
                          [30, 40]])

print("0D:", tensor_0d.dim())
print("1D:", tensor_1d.dim())
print("2D:", tensor_2d.dim())
```

## **11. Tensor Reshaping**

**Understand the Concept**

- Tensor reshaping changes the shape of a tensor.
- The total number of elements must remain the same.
- Reshaping is useful when the data shape needs to match a neural network layer.
- PyTorch provides functions like `reshape()` and `view()` for this.

## **12. `reshape()`**

**Understand the Concept**

- `reshape()` is used to change the shape of a tensor.
- It does not change the values in the tensor.
- The new shape must have the same total number of elements.

```python
print("Original shape:", data.shape)
```

```python
# Reshape tensor
new_data = data.reshape(3, 2)
print("New shape:", new_data.shape)
print(new_data)
```

## **13. `view()`**

**Understand the Concept**

- `view()` is used to change the shape of a tensor.
- It works similar to `reshape()`.
- The total number of elements must remain the same.
- The tensor should be stored in a suitable memory layout for `view()` to work.


```python
# Change shape
new_data = data.view(3, 2)
print("Original shape:", data.shape)
print("New shape:", new_data.shape)
print(new_data)
```

## **15. Tensor Arithmetic**

**Understand the Concept**

- Tensor arithmetic means performing mathematical operations on tensors.
- We can add, subtract, multiply, and divide tensors.
- These operations are used in neural network calculations.

**How It Works**

- PyTorch performs the operation element by element when the tensor shapes match.

## **16. Matrix Multiplication**

**Understand the Concept**

- Matrix multiplication is used to multiply two tensors based on their rows and columns.
- It is very important in neural network calculations.
- It is different from normal element-by-element multiplication.

**How It Works**

- The number of columns in the first matrix must match the number of rows in the second matrix.
- PyTorch uses `torch.matmul()` or `@` for matrix multiplication.

**Example**

- A `(2, 3)` tensor multiplied by a `(3, 2)` tensor produces a `(2, 2)` tensor.

```python
import torch

# Create matrices
a = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])

b = torch.tensor([[1, 2],
                  [3, 4],
                  [5, 6]])
```

```python
# Matrix multiplication
result = torch.matmul(a, b)
print("Result:")
print(result)
print("Shape:", result.shape)
```

## **17. Broadcasting**

**Understand the Concept**

- Broadcasting allows tensors with different shapes to work together.
- PyTorch automatically expands the smaller tensor when the shapes are compatible.
- It is useful for performing calculations without manually changing the tensor shape.

**How It Works**

- PyTorch checks the shapes of the tensors.
- If the shapes are compatible, it automatically broadcasts the smaller tensor.
- The calculation is then performed element by element.

**Example**

- A `(2, 3)` tensor can be added to a `(3,)` tensor.
- The `(3,)` tensor is applied to each row.

```python
# Broadcasting
# Create tensors
a = torch.tensor([[10, 20, 30],
                  [40, 50, 60]])
b = torch.tensor([1, 2, 3])

# Broadcasting
result = a + b
print("Result:")
print(result)
```

## **19. Device Management**

**Understand the Concept**

- Device management means choosing where PyTorch will run the calculations.
- We can use the CPU or GPU.
- The same model and tensors should be placed on the same device.

**How It Works**

- First, we check if a GPU is available.
- If a GPU is available, we use it.
- Otherwise, we use the CPU.

**Example**

- `cuda` → GPU
- `cpu` → CPU

```python
# Check GPU
if torch.cuda.is_available():
    print("GPU is available")
else:
    print("GPU is not available")
```

`torch.cuda.is_available()` checks whether a CUDA-enabled GPU is available.

```python
import torch

# Select device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)

# Create tensor on selected device
data = torch.tensor([10, 20, 30]).to(device)

print("Tensor:", data)
```

**Explanation**

- `torch.device()` selects the CPU or GPU.
- `.to(device)` moves the tensor to the selected device.
- This allows the same code to work with or without a GPU.

## **20. Converting NumPy ↔ PyTorch**

**Understand the Concept**

- We can convert a NumPy array into a PyTorch tensor.
- We can also convert a PyTorch tensor into a NumPy array.
- This is useful when we use both NumPy and PyTorch in the same project.

**How It Works**

- NumPy → PyTorch: `torch.from_numpy()`
- PyTorch → NumPy: `.numpy()`


```python
# NumPy to PyTorch
numpy_data = np.array([10, 20, 30])
tensor_data = torch.from_numpy(numpy_data)
print("NumPy:", numpy_data)
print("PyTorch:", tensor_data)
```

```python
# PyTorch to NumPy
new_numpy_data = tensor_data.numpy()
print("Back to NumPy:", new_numpy_data)
```

## **PyTorch Tensor Concepts**

| Concept | What is it? | Why is it required? | When to use it? |
|---|---|---|---|
| `torch.Tensor` | Main data structure used in PyTorch to store numerical data. | Used to store inputs, outputs, weights, and other values for neural networks. | Whenever we work with data in PyTorch. |
| `requires_grad` | Tells PyTorch to track operations on a tensor for gradient calculation. | Required during training to calculate how the loss changes with respect to model parameters. | Mainly used for tensors involved in learning and weight updates. |
| `torch.no_grad()` | Temporarily stops gradient calculation. | Saves memory and reduces unnecessary calculations. | Mainly during prediction or model evaluation when we don't need gradients. |

```python
# Gradient tracking
x = torch.tensor([2.0], requires_grad=True)
y = x ** 2
print(y)
```

```python
# No gradient tracking
with torch.no_grad():
    result = x ** 2
print(result)
```

### **Explanation**

- `torch.Tensor` stores the data used by PyTorch.
- `requires_grad=True` allows PyTorch to calculate gradients.
- `torch.no_grad()` stops gradient tracking when we only need predictions.
