# **Notebook 1 – Introduction to Deep Learning**

## **1. What is Deep Learning?**

**Understand the Concept**

- Deep Learning is a part of Machine Learning.
- It uses neural networks to learn patterns from data.
- It can work with large amounts of data.
- It is commonly used for images, text, speech, and other complex data.

**How It Works**

- We give data to a neural network.
- The network finds patterns in the data.
- It learns by changing its weights and reducing the error.
- After training, it can use what it learned to make predictions.



## **2. AI vs ML vs Deep Learning**

**Understand the Concept**

- **AI** is the bigger concept where machines perform tasks that normally need human intelligence.
- **Machine Learning** is a part of AI where machines learn from data.
- **Deep Learning** is a part of Machine Learning that uses neural networks with multiple layers.

**How It Works**

- AI → Machines perform intelligent tasks.
- ML → Machines learn patterns from data.
- Deep Learning → Neural networks learn complex patterns from large amounts of data.

**Example**

- AI → A chatbot answering questions.
- ML → Predicting whether a customer will churn.
- Deep Learning → Using a neural network to predict customer churn from many features.

## **3. Traditional ML vs Deep Learning**

**Understand the Concept**

- Traditional Machine Learning usually needs us to select and prepare useful features.
- Deep Learning can learn useful features automatically from the data.
- Traditional ML works well with smaller and structured datasets.
- Deep Learning is useful for large and complex data.

**How It Works**

- Traditional ML → Data → Feature Selection → ML Model → Prediction
- Deep Learning → Data → Neural Network → Prediction
- Deep Learning uses multiple layers to learn patterns step by step.

**Example**

- Traditional ML → Use customer tenure, contract type, and monthly charges to predict churn.
- Deep Learning → Give these features to a neural network and let it learn the patterns automatically.

## **4. Why Neural Networks Are Required**

**Understand the Concept**

- Neural networks are useful for learning complex patterns from data.
- They can learn relationships between many input features.
- They are useful when simple ML models may not capture complex patterns easily.

**How It Works**

- Neural networks take input data and pass it through different layers.
- Each layer learns some patterns from the data.
- The final layer gives the prediction.


## **5. Applications of Deep Learning**

**Understand the Concept**

- Deep Learning is used in many real-world applications.
- It is useful for working with images, text, speech, and complex data.

**Example**

- Image recognition
- Speech recognition
- Chatbots
- Recommendation systems
- Medical image analysis
- Customer churn prediction

## **6. Supervised Deep Learning**

**Understand the Concept**

- Supervised Deep Learning learns from data that has input and correct output.
- The model learns the relationship between the input and target.
- It is mainly used for classification and regression problems.

**How It Works**

- Give the model input data and the correct target.
- The model makes a prediction.
- The prediction is compared with the actual target.
- The model learns by reducing the error.



## **7. Unsupervised Deep Learning**

**Understand the Concept**

- Unsupervised Deep Learning works with data without a target or correct output.
- The model tries to find patterns or groups in the data.
- It is useful when we do not have labeled data.

**How It Works**

- Give the model input data without target labels.
- The model looks for hidden patterns or relationships.
- The learned patterns can be used for grouping or other analysis.


## **8. Neural Network Terminology**

**Understand the Concept**

- A neural network is made up of different parts that work together.
- The main terms are neuron, input, weight, bias, activation, and layer.
- These parts help the network learn patterns from data.

**How It Works**

- **Neuron** → Performs a calculation on the input.
- **Input** → Data given to the network.
- **Weight** → Shows how important an input is.
- **Bias** → Helps adjust the output.
- **Activation** → Decides the output of a neuron.
- **Layer** → A group of neurons working together.



## **9. Neuron**

**Understand the Concept**

- A neuron is the basic building block of a neural network.
- It takes input values and performs a calculation.
- It uses weights and bias to calculate the output.
- An activation function is then applied to the result.



**Formula**

- z = w1x1 + w2x2 + ... + b
- output = activation(z)

## **10. Input**

**Understand the Concept**

- Input is the data given to a neural network.
- Each input represents a feature or value.
- The neural network uses these inputs to make a prediction.



## **11. Weight**

**Understand the Concept**

- Weight shows how important an input is to a neuron.
- Each input has its own weight.
- During training, the model changes the weights to reduce the error.


## **12. Bias**

**Understand the Concept**

- Bias is an extra value added to the weighted sum.
- It helps the neuron adjust its output.
- The model also learns the bias during training.

**How It Works**

- First, the inputs are multiplied by their weights.
- Then the values are added together.
- Finally, the bias is added.

**Formula**

`z = w1x1 + w2x2 + b`



## **13. Activation**

**Understand the Concept**

- Activation is a function applied to the neuron's result.
- It helps the neural network learn complex patterns.
- Different activation functions are used for different problems.


**Example**

- ReLU is commonly used in hidden layers.
- Sigmoid can be used to get a probability for binary classification.

## **14. Layer**

**Understand the Concept**

- A layer contains one or more neurons.
- Neurons in the same layer process the input at the same stage.
- A neural network can have multiple layers.

**How It Works**

- The input is passed to a layer.
- The neurons process the values.
- The result is passed to the next layer.

**Example**

- A neural network can have an input layer, hidden layers, and an output layer.

## **15. Input Layer**

**Understand the Concept**

- The input layer is the first layer of a neural network.
- It receives the input features.
- It passes these values to the hidden layer.

**How It Works**

- Each input feature is given to the input layer.
- The values are passed to the neurons in the next layer.
- The network then continues processing the data.


## **16. Hidden Layer**

**Understand the Concept**

- A hidden layer is a layer between the input and output layers.
- It contains neurons that learn patterns from the input data.
- A neural network can have one or more hidden layers.

**How It Works**

- The hidden layer receives values from the input layer.
- Its neurons process the values using weights, bias, and activation.
- The result is passed to the next layer.



## **17. Output Layer**

**Understand the Concept**

- The output layer is the final layer of a neural network.
- It gives the final prediction.
- The number of output neurons depends on the problem.

**How It Works**

- The output layer receives values from the previous layer.
- It processes them and produces the final result.


## **18. Parameters**

**Understand the Concept**

- Parameters are values that the model learns during training.
- The main parameters are weights and biases.
- They are updated to reduce the model's error.

**How It Works**

- The model starts with initial weights and biases.
- During training, they are updated using gradients.
- The model continues this process until it learns useful patterns.


## **19. Hyperparameters**

**Understand the Concept**

- Hyperparameters are settings chosen before training.
- The model does not learn these values automatically.
- They control how the model is trained.

**How It Works**

- We choose the hyperparameters before training.
- The model uses them during the training process.

**Example**

- Learning rate
- Batch size
- Number of epochs
- Number of hidden layers
- Number of neurons

## **20. Tensor**

**Understand the Concept**

- A tensor is a data structure used to store numerical data.
- It can have one or more dimensions.
- PyTorch uses tensors to store and process data.

**How It Works**

- Data is stored inside tensors.
- Neural networks perform calculations using these tensors.
- Tensors can also be used with a GPU for faster computation.



## **21. Model**

**Understand the Concept**

- A model is the complete system that learns from data.
- A neural network model contains layers, weights, biases, and activation functions.
- After training, the model can make predictions on new data.

**How It Works**

- Data is given to the model.
- The model processes the data through its layers.
- It learns during training and then uses the learned parameters to make predictions.
