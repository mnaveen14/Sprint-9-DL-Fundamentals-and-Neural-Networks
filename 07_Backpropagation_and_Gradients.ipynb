# **Q1. Gradient**

**Understand the Concept**

- A gradient tells us how a value changes when a parameter changes.
- In neural networks, gradients tell us how the loss changes with respect to weights and biases.
- They help the model know which direction to move its parameters.

```python
import torch

x = torch.tensor(3.0, requires_grad=True)
y = x ** 2

y.backward()

print("x:", x.item())
print("y:", y.item())
print("Gradient dy/dx:", x.grad.item())
```

# **Explanation**

- Here `y = x²`.
- The gradient is `dy/dx = 2x`.
- At `x = 3`, the gradient is 6.
- PyTorch calculates it automatically after `.backward()`.

# **Q2. Why Gradients Are Required**

**Understand the Concept**

- Neural networks have many learnable parameters.
- We need to know how each parameter affects the loss.
- Gradients provide this information.
- The optimizer uses gradients to update the parameters.

```python
import torch

w = torch.tensor(2.0, requires_grad=True)
loss = (w - 5) ** 2

loss.backward()

print("Loss:", loss.item())
print("Gradient:", w.grad.item())
```

# **Explanation**

- The gradient tells us how the loss changes when `w` changes.
- The training algorithm can use this value to move `w` toward a better value.

# **Q3. Gradient Descent**

**Understand the Concept**

- Gradient Descent is an optimization method.
- It updates parameters in the opposite direction of the gradient.

**Formula**

`new weight = old weight - learning rate × gradient`

**Why?**

- The gradient points toward increasing loss.
- Moving in the opposite direction helps reduce the loss.

```python
import numpy as np

w = 2.0
gradient = 4.0
learning_rate = 0.1

new_w = w - learning_rate * gradient

print("Old weight:", w)
print("New weight:", new_w)
```

# **Explanation**

- The old weight is 2.
- The gradient is 4.
- With learning rate 0.1, the weight moves from 2 to 1.6.
- This is one gradient-descent update.

# **Q4. Computational Graph**

**Understand the Concept**

- A computational graph represents calculations as connected operations.
- PyTorch builds this graph when tensors require gradients.
- The graph is used during backpropagation.

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

a = x * 3
b = a + 2
y = b ** 2

y.backward()

print("Output:", y.item())
print("Gradient:", x.grad.item())
```

# **Explanation**

The calculation is:

`x → multiply by 3 → add 2 → square → y`

PyTorch tracks these operations and uses them to calculate the gradient of `y` with respect to `x`.

# **Q5. Forward Pass**

**Understand the Concept**

- The forward pass sends input through the model.
- The model calculates its prediction.
- The prediction is then compared with the target using a loss function.

**Flow**

`Input → Model → Prediction → Loss`

```python
import torch
import torch.nn as nn

model = nn.Linear(2, 1)

x = torch.tensor([[2.0, 3.0]])
y = torch.tensor([[5.0]])

prediction = model(x)
loss = nn.MSELoss()(prediction, y)

print("Prediction:", prediction.item())
print("Loss:", loss.item())
```

# **Explanation**

- The input passes through the model.
- The model produces a prediction.
- The loss function compares prediction and target.

# **Q6. Loss Calculation**

**Understand the Concept**

- Loss measures the model's error.
- The loss value becomes the starting point for backpropagation.
- Training tries to reduce this value.

```python
import torch
import torch.nn as nn

prediction = torch.tensor([[4.0]])
target = torch.tensor([[5.0]])

loss = nn.MSELoss()(prediction, target)

print("Loss:", loss.item())
```

# **Explanation**

- The prediction is 4 and the target is 5.
- MSE calculates the error.
- This loss is used to calculate gradients during the backward pass.

# **Q7. Backward Pass**

**Understand the Concept**

- The backward pass calculates gradients.
- It starts from the loss and moves backward through the computational graph.
- PyTorch performs this using `.backward()`.

```python
import torch

x = torch.tensor(2.0, requires_grad=True)
loss = (x - 5) ** 2

loss.backward()

print("Loss:", loss.item())
print("Gradient:", x.grad.item())
```

# **Explanation**

- `.backward()` starts the backward pass.
- PyTorch calculates how the loss changes with respect to `x`.
- The result is stored in `x.grad`.

# **Q8. Chain Rule**

**Understand the Concept**

- Neural networks contain many connected calculations.
- The chain rule helps calculate how an early value affects the final loss.
- Backpropagation uses the chain rule.

**Simple Example**

If:

`y = (3x + 2)²`

then the derivative is calculated through the connected operations.

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = (3 * x + 2) ** 2
y.backward()

print("y:", y.item())
print("dy/dx:", x.grad.item())
```

# **Explanation**

- The expression contains multiple operations.
- PyTorch applies the chain rule through the computational graph.
- `.grad` stores the final derivative.

# **Q9. Backpropagation**

**Understand the Concept**

Backpropagation is the process of calculating gradients by moving from the loss back through the network.

**Basic Flow**

`Forward Pass → Calculate Loss → Backward Pass → Calculate Gradients → Update Weights`

```python
import torch
import torch.nn as nn

model = nn.Linear(1, 1)

x = torch.tensor([[2.0]])
y = torch.tensor([[5.0]])

prediction = model(x)
loss = nn.MSELoss()(prediction, y)

loss.backward()

print("Loss:", loss.item())
print("Weight gradient:", model.weight.grad)
print("Bias gradient:", model.bias.grad)
```

# **Explanation**

- The model first makes a prediction.
- Loss is calculated.
- `.backward()` calculates gradients for the weight and bias.
- These gradients can then be used by an optimizer.

# **Q10. `requires_grad=True`**

**Understand the Concept**

- `requires_grad=True` tells PyTorch to track operations on a tensor.
- This is required when we want gradients for that tensor.

```python
import torch

x = torch.tensor(4.0, requires_grad=True)

y = x ** 2
y.backward()

print("x:", x.item())
print("Gradient:", x.grad.item())
```

# **Explanation**

- PyTorch tracks the calculation because `requires_grad=True`.
- After `.backward()`, the gradient is available in `x.grad`.

# **Q11. `.backward()`**

**Understand the Concept**

- `.backward()` calculates gradients.
- It is normally called on the loss during training.
- The calculated gradients are stored in the `.grad` attributes of tracked parameters.

```python
import torch

x = torch.tensor(3.0, requires_grad=True)
loss = x ** 2

loss.backward()

print("Gradient:", x.grad.item())
```

# **Explanation**

- `loss.backward()` starts gradient calculation.
- The gradient is stored in `x.grad`.

# **Q12. `.grad`**

**Understand the Concept**

- `.grad` stores the gradient calculated for a tensor.
- Optimizers use parameter gradients to update model weights.

```python
import torch

x = torch.tensor(3.0, requires_grad=True)
loss = x ** 2

loss.backward()

print("Stored gradient:", x.grad)
```

# **Explanation**

- After `.backward()`, `x.grad` contains the derivative.
- Before a backward pass, the gradient may be `None`.

# **Q13. PyTorch Autograd**

**Understand the Concept**

- Autograd is PyTorch's automatic differentiation system.
- It tracks operations and calculates gradients automatically.
- This avoids manually calculating derivatives for every model parameter.

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = x ** 3 + 2 * x

y.backward()

print("y:", y.item())
print("dy/dx:", x.grad.item())
```

# **Explanation**

- Autograd tracks the calculation of `y`.
- `.backward()` calculates the derivative automatically.
- This becomes very useful when neural networks contain thousands or millions of parameters.
