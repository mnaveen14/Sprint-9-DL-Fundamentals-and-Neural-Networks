# **Q1. Gradient Descent**

**Understand the Concept**

- Gradient Descent reduces the loss by updating model parameters.
- It moves parameters in the opposite direction of the gradient.

**Formula**

`w_new = w_old - learning_rate × gradient`

**Advantages**

- Simple.
- Easy to understand.

**Limitation**

- The learning rate must be chosen carefully.

```python
w = 5.0
gradient = 6.0
learning_rate = 0.1

w_new = w - learning_rate * gradient

print("Old weight:", w)
print("New weight:", w_new)
```

# **Explanation**

- The weight is changed using the gradient and learning rate.
- Repeating this process can reduce the loss.

# **Q2. Batch Gradient Descent**

**Understand the Concept**

- Batch Gradient Descent calculates the gradient using the entire training dataset.
- One update is made after processing all training samples.

**Advantage**

- The update uses information from the complete dataset.

**Limitation**

- Can be slow and memory-heavy for large datasets.

```python
import numpy as np

X = np.array([1.0, 2.0, 3.0])
y = np.array([2.0, 4.0, 6.0])
w = 1.0

pred = w * X
gradient = np.mean(2 * X * (pred - y))

w_new = w - 0.1 * gradient

print("Gradient:", gradient)
print("Updated weight:", w_new)
```

# **Explanation**

- All training samples are used before the weight update.
- This is the basic idea of batch gradient descent.

# **Q3. Stochastic Gradient Descent**

**Understand the Concept**

- Stochastic Gradient Descent (SGD) updates the model using one training sample at a time.
- It can make frequent updates.

**Advantages**

- Faster updates.
- Can work well with large datasets.

**Limitations**

- Updates can be noisy.
- The loss may fluctuate.

```python
import numpy as np

x = 2.0
y = 4.0
w = 1.0

pred = w * x
gradient = 2 * x * (pred - y)

w_new = w - 0.1 * gradient

print("Gradient:", gradient)
print("Updated weight:", w_new)
```

# **Explanation**

- Only one sample is used for this update.
- The weight is immediately updated.
- Repeating this for many samples gives SGD training.

# **Q4. Mini-batch Gradient Descent**

**Understand the Concept**

- Mini-batch Gradient Descent uses a small group of samples for each update.
- It is between batch and stochastic gradient descent.

**Example**

If the dataset has 1000 samples and batch size is 32, the model processes around 32 samples per update.

**Advantages**

- More efficient than using one sample.
- Uses less memory than the full dataset.
- Common in deep learning.

```python
import torch

X = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

batch = X[:2]
target = y[:2]

print("Mini-batch:")
print(batch)
print("Targets:")
print(target)
```

# **Explanation**

- Only a small group of samples is used for one update.
- Deep learning models commonly train using mini-batches.

# **Q5. Learning Rate**

**Understand the Concept**

- Learning rate controls how much the parameters change during each update.
- A very small learning rate can make training slow.
- A very large learning rate can make training unstable.

**Formula**

`new weight = old weight - learning rate × gradient`

```python
gradient = 5.0
w = 10.0

for lr in [0.001, 0.01, 0.1]:
    new_w = w - lr * gradient
    print("Learning rate:", lr, "New weight:", new_w)
```

# **Explanation**

- A smaller learning rate makes a smaller step.
- A larger learning rate makes a larger step.
- Choosing a suitable learning rate is important for training.

# **Q6. SGD Optimizer**

**Understand the Concept**

- PyTorch provides SGD through `torch.optim.SGD`.
- It updates parameters using their gradients.

```python
import torch
import torch.nn as nn

model = nn.Linear(1, 1)

optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

x = torch.tensor([[2.0]])
y = torch.tensor([[5.0]])

loss = nn.MSELoss()(model(x), y)

optimizer.zero_grad()
loss.backward()
optimizer.step()

print("Updated weight:", model.weight.data)
print("Updated bias:", model.bias.data)
```

# **Explanation**

- `zero_grad()` clears old gradients.
- `backward()` calculates new gradients.
- `step()` updates the model parameters using SGD.

# **Q7. SGD with Momentum**

**Understand the Concept**

- Momentum uses information from previous updates.
- It can help the optimizer move more smoothly.
- It can reduce some of the zig-zag movement of basic SGD.

**Advantages**

- Faster movement in useful directions.
- Can make training smoother.

**Limitation**

- Adds another hyperparameter.

```python
import torch
import torch.nn as nn

model = nn.Linear(1, 1)

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01,
    momentum=0.9
)

print(optimizer)
```

# **Explanation**

- `momentum=0.9` tells SGD to use previous update information.
- This can help training move more consistently toward a better solution.

# **Q8. RMSprop**

**Understand the Concept**

- RMSprop adapts the learning rate for different parameters.
- It uses a moving average of squared gradients.
- This can help when gradients have different scales.

**Advantages**

- Adaptive learning rates.
- Can work well for some neural-network problems.

**Limitation**

- Has extra hyperparameters to tune.

```python
import torch
import torch.nn as nn

model = nn.Linear(1, 1)

optimizer = torch.optim.RMSprop(
    model.parameters(),
    lr=0.001
)

print(optimizer)
```

# **Explanation**

- RMSprop adjusts parameter updates using recent gradient information.
- `lr` controls the base learning rate.

# **Q9. Adam**

**Understand the Concept**

- Adam combines ideas from momentum and adaptive learning rates.
- It is widely used in deep learning.
- It usually works well with less manual tuning than basic SGD.

**Advantages**

- Adaptive learning rate.
- Uses momentum-like information.
- Common and effective.

**Limitations**

- Can use more memory than basic SGD.
- SGD can sometimes generalize better in some problems.

```python
import torch
import torch.nn as nn

model = nn.Linear(1, 1)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

print(optimizer)
```

# **Explanation**

- Adam keeps running information about gradients.
- It uses that information to adapt parameter updates.
- `lr` controls the base learning rate.

# **Q10. Compare SGD vs Adam**

**Understand the Concept**

| Feature | SGD | Adam |
|---|---|---|
| Basic idea | Gradient-based update | Adaptive gradient update |
| Momentum | Optional | Built in |
| Learning rate | Usually chosen manually | Adapted per parameter |
| Memory | Lower | Higher |
| Common use | Strong baseline | Fast deep-learning training |

**When to Use**

- Start with Adam when you want a simple and strong baseline.
- Try SGD when you want to compare optimizers or tune for generalization.

```python
import torch
import torch.nn as nn

model_sgd = nn.Linear(2, 1)
model_adam = nn.Linear(2, 1)

sgd = torch.optim.SGD(model_sgd.parameters(), lr=0.01)
adam = torch.optim.Adam(model_adam.parameters(), lr=0.001)

print("SGD:", sgd)
print("\nAdam:", adam)
```

# **Explanation**

- Both optimizers update model parameters using gradients.
- Adam adapts updates using gradient statistics.
- SGD is simpler and uses less optimizer state.

# **Q11. Learning Rate Experiment**

**Understand the Concept**

- Different learning rates can produce different training behavior.
- A useful experiment is to compare a small, medium, and large learning rate.
- We can observe the loss during training.

```python
import torch
import torch.nn as nn

x = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

for lr in [0.001, 0.01, 0.1]:
    model = nn.Linear(1, 1)
    optimizer = torch.optim.SGD(model.parameters(), lr=lr)
    loss_fn = nn.MSELoss()

    for _ in range(100):
        prediction = model(x)
        loss = loss_fn(prediction, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print("Learning rate:", lr, "Final loss:", loss.item())
```

# **Explanation**

- The same model is trained with three learning rates.
- The final losses can be compared.
- This shows why learning rate is an important hyperparameter.
