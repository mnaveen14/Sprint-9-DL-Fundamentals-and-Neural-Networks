# **Q1. Why Loss Functions Are Required**

**Understand the Concept**

- A loss function measures how different the prediction is from the actual value.
- The model uses the loss to learn.
- Smaller loss usually means the prediction is closer to the target.

**Learning Flow**

`Prediction → Loss → Backpropagation → Weight Update`

```python
import torch

actual = torch.tensor([1.0])
prediction = torch.tensor([0.7])

loss = (actual - prediction) ** 2

print("Loss:", loss.item())
```

# **Explanation**

- The prediction is 0.7 and the actual value is 1.0.
- The difference is measured using a loss calculation.
- The training process uses this loss to improve the model.

# **Q2. MAE**

**Understand the Concept**

- MAE means Mean Absolute Error.
- It calculates the average absolute difference between actual and predicted values.

**Formula**

`MAE = (1/n) Σ |y - ŷ|`

**When to Use**

- Regression problems.
- Useful when you want an easy-to-understand error in the same unit as the target.

**Advantages**

- Simple.
- Less affected by large errors than MSE.

**Limitations**

- It does not strongly punish large errors.

```python
import numpy as np

actual = np.array([10, 20, 30])
predicted = np.array([12, 18, 27])

mae = np.mean(np.abs(actual - predicted))

print("MAE:", mae)
```

# **Explanation**

- Absolute error is calculated for each prediction.
- The errors are averaged.
- Lower MAE means predictions are closer to the actual values.

# **Q3. MSE**

**Understand the Concept**

- MSE means Mean Squared Error.
- It squares each error before taking the average.

**Formula**

`MSE = (1/n) Σ(y - ŷ)²`

**When to Use**

- Common for regression.
- Useful when large errors should be punished more.

**Advantages**

- Strongly penalizes large errors.
- Easy to optimize mathematically.

**Limitations**

- Sensitive to outliers.

```python
import numpy as np

actual = np.array([10, 20, 30])
predicted = np.array([12, 18, 27])

mse = np.mean((actual - predicted) ** 2)

print("MSE:", mse)
```

# **Explanation**

- Each error is squared.
- Squaring makes large errors more important.
- The average squared error is the MSE.

# **Q4. RMSE**

**Understand the Concept**

- RMSE means Root Mean Squared Error.
- It is the square root of MSE.

**Formula**

`RMSE = √MSE`

**When to Use**

- Regression problems.
- Useful when you want the error in the same unit as the target.

**Advantages**

- Same unit as the target.
- Large errors receive more importance.

**Limitations**

- Can be affected by outliers.

```python
import numpy as np

actual = np.array([10, 20, 30])
predicted = np.array([12, 18, 27])

mse = np.mean((actual - predicted) ** 2)
rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

# **Explanation**

- First, MSE is calculated.
- Then its square root is taken.
- RMSE is easier to interpret because it uses the original target unit.

# **Q5. Binary Cross Entropy**

**Understand the Concept**

- Binary Cross Entropy (BCE) is used for binary classification.
- It compares the actual binary label with the predicted probability.

**Formula**

`BCE = -[y log(p) + (1-y) log(1-p)]`

**Interpretation**

- A correct confident prediction gives a small loss.
- A wrong confident prediction gives a large loss.

```python
import numpy as np

y = 1
p = 0.8

bce = -(y * np.log(p) + (1 - y) * np.log(1 - p))

print("BCE:", bce)
```

# **Explanation**

- The actual class is 1.
- The predicted probability is 0.8.
- Because the prediction is reasonably close to 1, the loss is small.

# **Q6. BCE with Logits**

**Understand the Concept**

- `BCEWithLogitsLoss` combines sigmoid and binary cross entropy.
- It expects raw model scores called logits.
- It is numerically more stable than manually applying sigmoid followed by BCE.

**When to Use**

- Binary classification with one output neuron.
- Use raw logits as the model output.

```python
import torch
import torch.nn as nn

logits = torch.tensor([1.5])
target = torch.tensor([1.0])

loss_fn = nn.BCEWithLogitsLoss()
loss = loss_fn(logits, target)

print("BCE with logits loss:", loss.item())
```

# **Explanation**

- The model output is a raw logit.
- `BCEWithLogitsLoss()` handles the sigmoid and BCE calculation together.
- This is the common PyTorch choice for binary classification.

# **Q7. Cross Entropy Loss**

**Understand the Concept**

- Cross Entropy Loss is commonly used for multiclass classification.
- It compares class scores with the correct class.
- PyTorch `CrossEntropyLoss` expects raw logits.

**When to Use**

- Classification with multiple classes.
- One output score is produced for each class.

```python
import torch
import torch.nn as nn

logits = torch.tensor([[2.0, 1.0, 0.1]])
target = torch.tensor([0])

loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(logits, target)

print("Cross Entropy Loss:", loss.item())
```

# **Explanation**

- There are three class scores.
- `target = 0` means the first class is correct.
- Cross Entropy gives a lower loss when the correct class has a high score.

# **Q8. PyTorch Loss Functions**

**Understand the Concept**

PyTorch provides ready-to-use loss functions.

- `nn.L1Loss()` → MAE
- `nn.MSELoss()` → MSE
- RMSE can be calculated from MSE.
- `nn.BCELoss()` → BCE when probabilities are already available.
- `nn.BCEWithLogitsLoss()` → BCE with raw logits.
- `nn.CrossEntropyLoss()` → multiclass classification.

```python
import torch
import torch.nn as nn

actual = torch.tensor([10.0, 20.0, 30.0])
predicted = torch.tensor([12.0, 18.0, 27.0])

mae = nn.L1Loss()(predicted, actual)
mse = nn.MSELoss()(predicted, actual)
rmse = torch.sqrt(mse)

print("MAE:", mae.item())
print("MSE:", mse.item())
print("RMSE:", rmse.item())
```

# **Explanation**

- PyTorch provides built-in loss classes.
- This avoids writing the full loss formula every time.
- The same loss functions can be used inside a training loop.

# **Q9. Compare Loss Functions**

**Understand the Concept**

| Problem | Loss Function | Main Use |
|---|---|---|
| Regression | MAE | Average absolute error |
| Regression | MSE | Penalize large errors |
| Regression | RMSE | Error in original target unit |
| Binary classification | BCE | Probability-based binary prediction |
| Binary classification | BCEWithLogitsLoss | Raw logits + BCE |
| Multiclass classification | CrossEntropyLoss | Multiple classes |

```python
import torch
import torch.nn as nn

print("Regression:")
print("MAE  ->", nn.L1Loss())
print("MSE  ->", nn.MSELoss())

print("\nBinary classification:")
print("BCEWithLogitsLoss ->", nn.BCEWithLogitsLoss())

print("\nMulticlass classification:")
print("CrossEntropyLoss ->", nn.CrossEntropyLoss())
```

# **Explanation**

- Choose the loss based on the type of prediction problem.
- MAE, MSE, and RMSE are used for regression.
- BCE-based losses are used for binary classification.
- Cross Entropy is commonly used for multiclass classification.

# **Q10. Prediction → Loss → Learning**

**Understand the Concept**

The basic learning idea is:

`Prediction → Loss → Learning`

- The model makes a prediction.
- Loss measures the error.
- Backpropagation calculates how parameters should change.
- The optimizer updates the parameters.

```python
import torch
import torch.nn as nn

model = nn.Linear(1, 1)

x = torch.tensor([[2.0]])
y = torch.tensor([[5.0]])

prediction = model(x)

loss_fn = nn.MSELoss()
loss = loss_fn(prediction, y)

print("Prediction:", prediction.item())
print("Loss:", loss.item())
```

# **Explanation**

- The model first produces a prediction.
- MSE measures the difference from the target.
- During training, this loss is then used for backpropagation and weight updates.
