# **Q1. Overfitting**

**Understand the Concept**

- Overfitting means the model learns training data too closely.
- Training performance can be very good while validation performance gets worse.

```python
train_loss = [0.8, 0.5, 0.3, 0.15, 0.08]
val_loss = [0.85, 0.60, 0.50, 0.65, 0.90]
print(train_loss)
print(val_loss)
```

## **Explanation**

- Training loss keeps falling while validation loss rises after a point.
- This is a common overfitting pattern.

# **Q2. Underfitting**

**Understand the Concept**

- Underfitting means the model has not learned enough useful patterns.
- Training and validation performance can both remain poor.

```python
train_acc = 0.65
val_acc = 0.63
print("Training:", train_acc)
print("Validation:", val_acc)
```

## **Explanation**

- Both scores are low, suggesting the model may be too simple or not trained enough.

# **Q3. Bias**

**Understand the Concept**

- High bias is often related to underfitting.
- A high-bias model may be too simple to capture important patterns.

```python
model_complexity = ["Low", "Medium", "High"]
for x in model_complexity:
    print(x)
```

## **Explanation**

- Very simple models can have high bias.
- More capacity can reduce bias, but too much capacity can increase overfitting.

# **Q4. Variance**

**Understand the Concept**

- High variance means a model is sensitive to the training data.
- It is commonly associated with overfitting.

```python
train_accuracy = 0.99
validation_accuracy = 0.78
print("Gap:", train_accuracy - validation_accuracy)
```

## **Explanation**

- A large train-validation gap can indicate high variance.

# **Q5. Generalization**

**Understand the Concept**

- Generalization means performing well on unseen data.
- A good model learns useful patterns rather than memorizing samples.

```python
train_accuracy = 0.92
validation_accuracy = 0.90
print("Accuracy gap:", train_accuracy - validation_accuracy)
```

## **Explanation**

# **Q6. Dropout**

**Understand the Concept**

- Dropout randomly disables some neurons during training.
- It reduces dependence on particular neurons.

```python
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(10, 32),
    nn.ReLU(),
    nn.Dropout(0.3),
    nn.Linear(32, 1)
)
print(model)
```

## **Explanation**

- `Dropout(0.3)` randomly drops 30% of activations during training.

# **Q7. L1 Regularization**

**Understand the Concept**

- L1 adds the absolute values of weights to the loss.
- It can encourage some weights toward zero.

```python
import torch
weights = torch.tensor([1.0, -2.0, 0.5])
l1 = 0.01 * torch.sum(torch.abs(weights))
print("L1 penalty:", l1.item())
```

## **Explanation**

- The absolute weights are summed and multiplied by the regularization strength.

# **Q8. L2 Regularization**

**Understand the Concept**

- L2 adds a penalty based on squared weights.
- It encourages smaller weights.

```python
import torch
weights = torch.tensor([1.0, -2.0, 0.5])
l2 = 0.01 * torch.sum(weights ** 2)
print("L2 penalty:", l2.item())
```

## **Explanation**

- Squaring makes large weights receive a larger penalty.

# **Q9. Weight Decay**

**Understand the Concept**

- Weight decay adds a regularization effect during optimization.
- PyTorch optimizers provide the `weight_decay` argument.

```python
import torch
import torch.nn as nn

model = nn.Linear(4, 1)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001, weight_decay=0.01)
print(optimizer)
```

## **Explanation**

- Weight decay can help reduce overfitting by discouraging overly large weights.

# **Q10. Batch Normalization**

**Understand the Concept**

- Batch Normalization normalizes intermediate activations.
- It can make training more stable and can also provide some regularization effect.

```python
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(8, 16),
    nn.BatchNorm1d(16),
    nn.ReLU(),
    nn.Linear(16, 1)
)
print(model)
```

## **Explanation**

- `BatchNorm1d` normalizes the hidden outputs before ReLU.

# **Q11. Early Stopping**

**Understand the Concept**

- Early stopping stops training when validation performance stops improving.
- It can reduce overfitting.

```python
best = float("inf")
wait = 0
patience = 2
val_losses = [0.6, 0.5, 0.45, 0.46, 0.48]

for epoch, loss in enumerate(val_losses, 1):
    if loss < best:
        best = loss
        wait = 0
    else:
        wait += 1
    print(epoch, loss)
    if wait >= patience:
        print("Early stopping triggered")
        break
```

## **Explanation**

- The best validation loss is saved.
- If it stops improving for the patience period, training stops.

# **Q12. Without vs With Regularization**

**Understand the Concept**

- Compare a network without dropout with one using dropout.
- The goal is to see whether validation performance improves.

```python
import torch.nn as nn

without_dropout = nn.Sequential(nn.Linear(10,32), nn.ReLU(), nn.Linear(32,1))
with_dropout = nn.Sequential(nn.Linear(10,32), nn.ReLU(), nn.Dropout(0.3), nn.Linear(32,1))

print("Without:", without_dropout)
print("With:", with_dropout)
```

## **Explanation**

# **Q13. Training and Validation Curves**

**Understand the Concept**

- Curves help identify overfitting.
- Compare training and validation loss rather than looking at training loss alone.

```python
import matplotlib.pyplot as plt

epochs = range(1, 11)
train = [0.8,0.6,0.45,0.32,0.22,0.15,0.10,0.07,0.05,0.04]
val = [0.82,0.65,0.52,0.48,0.50,0.55,0.62,0.70,0.78,0.85]
plt.plot(epochs, train, label="Training")
plt.plot(epochs, val, label="Validation")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Training vs Validation Loss")
plt.legend(); plt.grid(); plt.show()
```

## **Explanation**

- Training loss decreases continuously.
- Validation loss starts increasing after a point, indicating overfitting.
