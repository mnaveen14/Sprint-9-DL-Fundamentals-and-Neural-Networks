# **Q1. Training Accuracy**

**Understand the Concept**

- Training accuracy measures performance on training data.
- It should generally improve as the model learns.

```python
history = [0.70, 0.78, 0.84, 0.88, 0.91]
for e, score in enumerate(history,1): print(f"Epoch {e}: {score:.2f}")
```

## **Explanation**

- High training accuracy alone does not guarantee good generalization.

# **Q2. Validation Accuracy**

**Understand the Concept**

- Validation accuracy measures performance on unseen validation data.
- It helps monitor generalization.

```python
history = [0.68, 0.75, 0.80, 0.81, 0.80]
for e, score in enumerate(history,1): print(f"Epoch {e}: {score:.2f}")
```

## **Explanation**

- If validation accuracy stops improving while training accuracy keeps rising, overfitting may be occurring.

# **Q3. Training Loss**

**Understand the Concept**

- Training loss measures error on training data.
- A decreasing curve usually means the model is learning.

```python
import matplotlib.pyplot as plt
loss = [0.8,0.6,0.45,0.32,0.22]
plt.plot(loss); plt.xlabel("Epoch"); plt.ylabel("Training Loss"); plt.title("Training Loss"); plt.grid(); plt.show()
```

## **Explanation**

- The decreasing curve shows improving training performance.

# **Q4. Validation Loss**

**Understand the Concept**

- Validation loss measures error on unseen validation data.
- It is useful for identifying overfitting.

```python
import matplotlib.pyplot as plt
loss = [0.82,0.64,0.50,0.52,0.60]
plt.plot(loss); plt.xlabel("Epoch"); plt.ylabel("Validation Loss"); plt.title("Validation Loss"); plt.grid(); plt.show()
```

## **Explanation**

- Validation loss decreases first and then rises, which can indicate overfitting.

# **Q5. Training vs Validation Curves**

**Understand the Concept**

- Comparing both curves gives more information than one curve alone.

```python
import matplotlib.pyplot as plt
e = range(1,6)
train=[0.8,0.6,0.45,0.32,0.22]
val=[0.82,0.64,0.50,0.52,0.60]
plt.plot(e,train,label="Training"); plt.plot(e,val,label="Validation")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.title("Training vs Validation Loss"); plt.legend(); plt.grid(); plt.show()
```

## **Explanation**

- Training loss keeps falling while validation loss rises after a point.
- This suggests overfitting.

# **Q6. Confusion Matrix**

**Understand the Concept**

- A confusion matrix shows correct and incorrect predictions for every class.

```python
import numpy as np
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
y_true=[0,0,1,1,2,2,2,1]
y_pred=[0,1,1,1,2,0,2,2]
cm=confusion_matrix(y_true,y_pred)
plt.imshow(cm); plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.title("Confusion Matrix"); plt.colorbar(); plt.show()
print(cm)
```

## **Explanation**

- Diagonal values are correct predictions.
- Off-diagonal values show errors.

# **Q7. Class-Wise Performance**

**Understand the Concept**

- Overall accuracy can hide weak classes.
- Class-wise scores show which classes need attention.

```python
scores = cm.diagonal()/cm.sum(axis=1)
for i,s in enumerate(scores): print(f"Class {i}: {s:.3f}")
```

## **Explanation**

- A low score for one class suggests a class-specific problem.

# **Q8. Misclassified Samples**

**Understand the Concept**

- Looking at incorrect samples helps explain model errors.
- Check whether samples are unclear, similar to another class, or incorrectly labeled.

```python
import numpy as np
import matplotlib.pyplot as plt
images=np.random.rand(5,28,28)
true=[0,1,2,1,0]; pred=[1,1,2,2,0]
for i in range(5):
    if true[i]!=pred[i]:
        plt.imshow(images[i],cmap="gray"); plt.title(f"True: {true[i]} | Pred: {pred[i]}"); plt.axis("off"); plt.show()
```

## **Explanation**

- Misclassified samples reveal errors that a single accuracy value cannot explain.

# **Q10. Overfitting**

**Understand the Concept**

- High training performance with lower validation performance can indicate overfitting.

```python
train_acc=0.97
val_acc=0.81
print("Gap:", train_acc-val_acc)
```

## **Explanation**

- A large train-validation gap is a warning sign.
- Dropout, early stopping, more data, or other changes may help.

# **Q11. Underfitting**

**Understand the Concept**

- Low training and validation performance can indicate underfitting.

```python
train_acc=0.65
val_acc=0.63
print(train_acc,val_acc)
```

## **Explanation**

- Both scores are low, suggesting the model has not learned enough useful patterns.

# **Q12. Class Confusion**

**Understand the Concept**

- Class confusion occurs when one class is frequently predicted as another.

```python
import numpy as np
cm=np.array([[90,10,0],[8,70,22],[0,18,82]])
for i in range(3):
    for j in range(3):
        if i!=j and cm[i,j]>10: print(f"Actual {i} -> Predicted {j}: {cm[i,j]}")
```

## **Explanation**

- Large off-diagonal values reveal repeated class confusion.

# **Q13. Poorly Performing Classes**

- What errors happen most often?
- Are problems data-related or model-related?
- What should be improved next?

```python
analysis={"training_accuracy":0.96,"validation_accuracy":0.84,"main_issue":"Possible overfitting","next_step":"Try dropout or early stopping"}
for k,v in analysis.items(): print(f"{k}: {v}")
```
