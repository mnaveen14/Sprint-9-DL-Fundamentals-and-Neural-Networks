# **Q1. Forward Propagation**

**Understand the Concept**

- Forward propagation means passing input data through a neural network.
- The network uses weights, biases, and activation functions.
- The final output becomes the prediction.

**Flow**

`Input → Hidden Layer → Output Layer`

**How It Works**

- Input enters the network.
- Each layer performs a calculation.
- Activation is applied.
- The final layer produces the prediction.

```python
import numpy as np

x = np.array([2.0, 3.0])

W1 = np.array([[0.5, 0.2],
               [0.3, 0.4]])
b1 = np.array([0.1, 0.2])

W2 = np.array([[0.6, 0.5]])
b2 = np.array([0.1])

z1 = np.dot(x, W1) + b1
a1 = np.maximum(0, z1)

z2 = np.dot(a1, W2.T) + b2
output = z2

print("Hidden linear output:", z1)
print("Hidden activation:", a1)
print("Final prediction:", output)
```

# **Explanation**

- `z1` calculates the weighted sum in the hidden layer.
- ReLU converts `z1` into `a1`.
- `z2` calculates the output layer value.
- The final value is the prediction.

# **Q2. Input Layer**

**Understand the Concept**

- The input layer receives the features.
- It does not normally perform a learning transformation by itself.
- The number of input values depends on the number of features.

**Example**

If a sample has:

`Age, Income, Tenure`

then the network has 3 input features.

```python
import torch
import torch.nn as nn

x = torch.tensor([[25.0, 50000.0, 12.0]])

layer = nn.Linear(3, 2)
output = layer(x)

print("Input shape:", x.shape)
print("Output shape:", output.shape)
```

# **Explanation**

- The input has 3 features.
- `nn.Linear(3, 2)` accepts 3 input features.
- It produces 2 values for the next layer.

# **Q3. Hidden Layer**

**Understand the Concept**

- Hidden layers are between the input and output layers.
- They learn patterns from the input features.
- Hidden layers normally use activation functions.

**Basic Calculation**

`z = xW + b`

`a = activation(z)`

```python
import torch
import torch.nn as nn

x = torch.tensor([[2.0, 3.0]])

hidden = nn.Linear(2, 3)
activation = nn.ReLU()

z = hidden(x)
a = activation(z)

print("Before activation:", z)
print("After activation:", a)
```

# **Explanation**

- The linear layer calculates the weighted sum plus bias.
- ReLU adds non-linearity.
- The result is passed to the next layer.

# **Q4. Output Layer**

**Understand the Concept**

- The output layer produces the final model output.
- Its size depends on the problem.
- Binary classification can use one output.
- Multiclass classification can use one output for each class.
- Regression can use one or more continuous outputs.

```python
import torch
import torch.nn as nn

x = torch.tensor([[2.0, 3.0]])

output_layer = nn.Linear(2, 1)
output = output_layer(x)

print("Output:", output)
```

# **Explanation**

- `nn.Linear(2, 1)` takes 2 features and produces 1 output.
- That single output can be used for a simple regression or binary-classification model.

# **Q5. Weights and Biases**

**Understand the Concept**

- Weights decide how strongly each input affects the output.
- Bias shifts the result.
- Both are learnable parameters.

**Formula**

`z = xW + b`

```python
import torch
import torch.nn as nn

layer = nn.Linear(2, 1)

print("Weights:")
print(layer.weight)

print("\nBias:")
print(layer.bias)
```

# **Explanation**

- `layer.weight` contains the weight values.
- `layer.bias` contains the bias value.
- During training, the optimizer updates these parameters.

# **Q6. Linear Transformation**

**Understand the Concept**

A neural-network layer first performs a linear transformation.

**Formula**

`z = xW + b`

This combines the input features using weights and bias.

```python
import numpy as np

x = np.array([2.0, 3.0])
W = np.array([0.5, 0.2])
b = 0.1

z = np.dot(x, W) + b

print("Linear transformation:", z)
```

# **Explanation**

- `np.dot(x, W)` calculates the weighted sum.
- `b` is added after the weighted sum.
- The result is the value before activation.

# **Q7. Activation**

**Understand the Concept**

- After the linear transformation, an activation function can be applied.
- Activation adds non-linearity.
- ReLU is commonly used in hidden layers.

**Formula**

`ReLU(x) = max(0, x)`

```python
import numpy as np

z = np.array([-2.0, 1.0, 3.0])

activation = np.maximum(0, z)

print("Before activation:", z)
print("After ReLU:", activation)
```

# **Explanation**

- Negative values become 0.
- Positive values stay unchanged.
- This activated output is passed to the next layer.

# **Q8. Prediction**

**Understand the Concept**

- Prediction is the final output produced by the network.
- The output interpretation depends on the task.
- For classification, the output can represent a class score or probability.
- For regression, it can be a continuous value.

```python
import torch
import torch.nn as nn

x = torch.tensor([[2.0, 3.0]])

model = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(),
    nn.Linear(4, 1)
)

prediction = model(x)

print("Prediction:", prediction)
```

# **Explanation**

- The input passes through the hidden layer.
- ReLU is applied.
- The output layer produces the final prediction.
- This complete process is forward propagation.

# **Q9. Manual NumPy Forward Propagation**

**Understand the Concept**

- A small neural network can be calculated manually.
- This helps us understand what happens inside a neural network.

**Network**

`Input → Hidden Layer → Output Layer`

We will calculate each step separately.

```python
import numpy as np

x = np.array([1.0, 2.0])

W1 = np.array([[0.5, 0.2],
               [0.4, 0.3]])
b1 = np.array([0.1, 0.1])

W2 = np.array([[0.7, 0.6]])
b2 = np.array([0.2])

z1 = np.dot(x, W1) + b1
a1 = np.maximum(0, z1)

z2 = np.dot(a1, W2.T) + b2

print("Step 1 - Hidden linear:", z1)
print("Step 2 - Hidden activation:", a1)
print("Step 3 - Output:", z2)
```

# **Explanation**

- First, the input is multiplied by the hidden-layer weights.
- The hidden bias is added.
- ReLU is applied.
- The activated values are passed to the output layer.
- The final value is the prediction.

# **Q10. Same Network Using PyTorch**

**Understand the Concept**

- PyTorch performs these layer calculations for us.
- `nn.Linear` stores the weights and biases.
- The `forward` process can be written using layers.

```python
import torch
import torch.nn as nn

class SimpleNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden = nn.Linear(2, 2)
        self.output = nn.Linear(2, 1)

    def forward(self, x):
        x = self.hidden(x)
        x = torch.relu(x)
        x = self.output(x)
        return x

model = SimpleNetwork()

x = torch.tensor([[1.0, 2.0]])
prediction = model(x)

print("Prediction:", prediction)
```

# **Explanation**

- `nn.Module` is the base class for the network.
- The hidden layer performs the first linear transformation.
- `torch.relu()` applies activation.
- The output layer produces the final prediction.
- This is the PyTorch version of forward propagation.
